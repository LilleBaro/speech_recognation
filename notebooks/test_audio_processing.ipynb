```python
import torch
import os
# os.add_dll_directory(r"C:\ffmpeg-8.1.2-full_build-shared")
import torchaudio
from torchaudio.transforms import Resample
# torchaudio.set_audio_backend("soundfile")
```

```python
def  load_audio(path):
    waveform, sample_rate = torchaudio.load(path, backend="soundfile")
    return waveform, sample_rate
```

```python
sample = load_audio("sample.mp3")
```

```python
sample
```

```python
def convert_to_mono(waveform):
    wave = waveform[0]
    if wave.shape[0] > 1:
        return torch.mean(waveform, dim=0, keepdim=True)
    return waveform
```

```python
conv = convert_to_mono(sample)
```

```python
def audio_resampling(audio):
  new_freq = 16000
  resampled_audio = Resample(orig_freq=sample[1], new_freq=new_freq, dtype=torch.float32)(audio)
  return resampled_audio, new_freq
```

```python
def normalize_audio(sample):
  normalized_audio = sample / torch.max(torch.abs(sample))
  return normalized_audio
```

```python
def preprocess_audio(audio):
  converted_audio = convert_to_mono(audio)
  resampled_audio, resampled_sample_rate = audio_resampling(converted_audio[0])
  normalized_audio = normalize_audio(resampled_audio)
  return normalized_audio, resampled_sample_rate
```

```python
processed_audio = preprocess_audio(sample)
processed_audio
```
